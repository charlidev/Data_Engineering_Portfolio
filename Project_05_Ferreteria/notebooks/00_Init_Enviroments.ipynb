```sql
-- Crear los Catálogos (Ambientes)
CREATE CATALOG IF NOT EXISTS ferreteria_dev;
CREATE CATALOG IF NOT EXISTS ferreteria_test;
CREATE CATALOG IF NOT EXISTS ferreteria_prod;

-- Crear los esquemas para DEV
USE CATALOG ferreteria_dev;
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- Crear los esquemas para test
USE CATALOG ferreteria_test;
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- Crear los esquemas para prod
USE CATALOG ferreteria_prod;
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- Otorgar permisos limitados al grupo de analistas en PROD
GRANT USE CATALOG ON CATALOG ferreteria_prod TO `analistas_ferreteria`;
GRANT USE SCHEMA ON SCHEMA ferreteria_prod.gold TO `analistas_ferreteria`;
GRANT SELECT ON SCHEMA ferreteria_prod.gold TO `analistas_ferreteria`;
```
